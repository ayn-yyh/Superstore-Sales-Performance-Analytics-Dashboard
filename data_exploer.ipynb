通过对销售数据进行描述性统计分析，数据集共包含9994条订单记录，涵盖2014年至2017年的销售业务数据。  
从销售指标来看，订单销售额（Sales）的平均值为229.86美元，中位数为54.49美元，二者存在较大差异，说明销售额呈现明显的右偏分布，少量高价值订单贡献了较高的销售金额。  
利润（Profit）的平均值为28.66美元，但最小值达到-6599.98美元，表明部分订单存在亏损情况。由于负利润可能反映折扣策略、产品成本或销售策略问题，因此在后续分析中不将其作为异常数据删除，而是进一步分析亏损订单的分布特征。  
折扣（Discount）的平均水平为15.62%，最高达到80%，说明不同订单之间存在较大的价格优惠差异。较高折扣可能促进销售增长，但也可能降低企业利润，因此后续将结合利润指标进一步分析折扣对盈利能力的影响。  
商品购买数量平均为3.78件，最大购买数量为14件，整体分布较为稳定，不存在明显异常情况。

```python
#利润率
df['Profit Margin'] = (df['Profit'] / df['Sales'])
```

```python
#订单年份
df['Order Year'] = (df['Order Date'].dt.year)
```

```python
#订单月份
df['Order Month'] = (df['Order Date'].dt.month)
```

```python
#季度
df['Order Quarter'] = (df['Order Date'].dt.quarter)
```

```python
#订单处理时间
df['Order Processing Days'] = ((df['Ship Date'] - df['Order Date']).dt.days)
```

### Date Warehouse Modeling

```python
#新增客户维度表
dim_customer = df[
    [
        'Customer ID',
        'Customer Name',
        'Segment',
        'Country',
        'State',
        'City',
        'Postal Code',
        'Region'
    ]
].drop_duplicates(
    subset=['Customer ID']
)
```

```python
dim_customer.head()
```

```python
dim_customer.shape
```

```python
#创建商品维度表
dim_product = df[
    ['Product ID',
     'Product Name',
     'Category',
     'Sub-Category'
    ]
].drop_duplicates(
    subset = ['Product ID']
)
```

```python
dim_product.head()
```

```python
dim_product.shape
```

```python
dim_product['Product ID'].nunique()
```

```python
#创建销售事实表
fact_sales = df[
[
'Row ID',
'Order ID',
'Customer ID',
'Product ID',
'Order Date',
'Sales',
'Quantity',
'Discount',
'Profit',
'Profit Margin',
'Order Processing Days'
]
].copy()
```

```python
# 修改字段名称，使其符合数据仓库命名规范

fact_sales.rename(
    columns={
        'Row ID': 'Sales ID'
    },
    inplace=True
)
```

```python
fact_sales.head()
```

```python
fact_sales.shape
```

```python
fact_sales.duplicated().sum()
```

```python
#创建日期维度表
date_range = pd.date_range(
    start=df['Order Date'].min(),
    end=df['Order Date'].max()
)


dim_date = pd.DataFrame(
    {
        'Date': date_range
    }
)
```

```python
dim_date.head()
```

```python
# 年份

dim_date['Year'] = (
    dim_date['Date'].dt.year
)


# 月份

dim_date['Month'] = (
    dim_date['Date'].dt.month
)


# 月份名称

dim_date['Month Name'] = (
    dim_date['Date'].dt.month_name()
)


# 季度

dim_date['Quarter'] = (
    dim_date['Date'].dt.quarter
)
```

```python
dim_date['Date ID'] = (
    dim_date['Date']
    .dt.strftime('%Y%m%d')
    .astype(int)
)
```
